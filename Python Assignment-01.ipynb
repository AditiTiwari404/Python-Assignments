```python
#List is a colection which is ordered and changable.
#It allows duplicate members
#Stores multiple items in 1 variable

list=["Rahul","Sakshi","Ankita"]
print(list)
```

```python
#Insert-adds an element at a special position

list.insert(2,"shreya")
print(list)
```

```python
#Append-adds an element at the end of the list

list.append("shreya")
print(list)
```

```python
#Remove-removes/deletes the specified item

list.remove("Sakshi")
print(list)
```

```python
#Pop-removes an element from a specied index

list.pop(1)
print(list)
```

```python
#Sort-Sorts the list in alphabetical order

list.sort()
print(list)
```

```python
#Tuple is a collection which is ordered and unchangable
#It allows duplicate members

x=["Rahul","Sakshi"]
print(x)
```

```python
#Length-To determine how many items a tuple has

x=["Rahul","Sakshi"]
print(len(x))
```

```python
#Append-Convert the tuple into a list then add element and then convert back to tuple

x=["Rahul","Sakshi"]
y=list(x)
y.append("Ankita")
x=tuple(y)
print(x)
```

```python
#Multiple-to multiply the context of tuple given times you can use (*) operator

fruits=["apple","banana"]
x=fruits*2
print(x)
```

```python
#Remove-convert into list then remove an item then convert back into tuple

x=["Rahul","22","Shreya"]
y=list(x)
y.remove("22")
x=tuple(y)
print(x)
```

```python
#Empty-to create an empty tuple

mytuple=()
print(mytuple)
```

```python
#dictionary is a collection which is ordered and changable.
#No duplicate entries
#Keys-This method will return a list of all keys in the dictionary
#get()-returns the value associated with a key
#Values-returns all values
#Items-returns all key-value parts

dictionary={"Rahul":1,"Sakshi":2,"Ankita":3}
print("keys:",dictionary.keys())
print("get value for 'Sakshi':",dictionary.get("Sakshi"))
print("values:",dictionary.values())
print("items:",dictionary.items())
```
